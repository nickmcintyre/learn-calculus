```python
width = 12
depth = 1
height = 3

# Display the net
print("  +" + "-" * width + "+")
print(" " * 2 + "|" + " " * width + "|")
print("+-+" + "-" * width + "+-+")
for i in range(height):
    print("| |" + " " * width + "| |")
print("+-+" + "-" * width + "+-+")
print(" " * 2 + "|" + " " * width + "|")
print("  +" + "-" * width + "+")
for i in range(height):
    print("  |" + " " * width + "|")
print("  +" + "-" * width + "+")

# Compute the surface area
area = 0

# Pair 1
area += 2 * width * depth

# Pair 2
area += 2 * width * height

# Pair 3
area += 2 * depth * height
print(f"The surface area is {area} square units.")
```

### 1-15

```python
# A ball bouncing off a wall
width = 5
x = 0
x_speed = 1
for i in range(10):
    clear_output(True)
    # Display the ball and wall
    print(" " * x + "o" + " " * (width - x) + "|")
    # Check for a collision
    if x > width - 1:
        x_speed = -1
    # Update position
    x += x_speed
    time.sleep(1)
    clear_output(True)
    print()
```

```python
width = 20
depth = 1
height = 6

pad = 11
num_rows = height - 2
for i in range(10):
    clear_output(True)
    print(" " * pad + "  +" + "-" * width + "+")
    print(" " * pad + " /|" + " " * (width - 1) + "/|")
    print(" " * pad + "+" + "-" * width + "+ |")
    for row in range(num_rows):
        if row < num_rows // 2:
            text = " " * pad + "| |" + " " * (width - 2) + "| |"
        if row > num_rows // 2 + 1:
            text = " " * pad + "| |" + " " * (width - 6) + " | " + " | |"
        if row == num_rows // 2:
            in_dashes = i % 5 + 1
            in_dash_pad = 5 - in_dashes
            out_dashes = i % 3 + 1
            out_dash_pad = 3 - out_dashes
            text = "***** " + "-" * in_dashes + " " * in_dash_pad + "|>|" + " " * (width - 2) + "|-|" + "-" * out_dashes + " " * out_dash_pad + " >**"
        if row == num_rows // 2 + 1:
            text = " " * pad + "| |  ^" + " " * (width - 10) + "****" + " | |"
        print(text)
    print(" " * pad + "| +" + "- | " + "-" * (width - 10) + " | " + "-|-+")
    print(" " * pad + "|/" + "   * " + " " * (width - 9) + "v" + "  |/ ")
    print(" " * pad + "+" + "-" * width + "+")
    time.sleep(1)
    clear_output(True)
    print()
```
