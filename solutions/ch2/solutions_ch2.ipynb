### 2-4

````python
showscreen()
clearscreen()

t = Turtle()

width = 100
height = 100
depth = 60
angle = 60

# Bottom and sides
t.forward(width)
t.left(angle)
t.forward(depth)
t.left(180 - angle)
t.forward(width)
t.left(angle)
t.forward(depth)
t.left(180 - angle)

# Top
y_bottom = t.ycor()
y_top = y_bottom + height
t.sety(y_top)
t.forward(width)
t.left(angle)
t.sety(y_bottom)
t.sety(y_top)
t.forward(depth)
t.left(180 - angle)
y_top = t.ycor()
y_bottom = y_top - height
t.sety(y_bottom)
t.sety(y_top)
t.forward(width)
t.left(angle)
y_top = t.ycor()
y_bottom = y_top - height
t.sety(y_bottom)
t.sety(y_top)
t.forward(depth)
t.left(180 - angle)

# Mass flux in
t.penup()
t.goto(-width / 2, height / 2)
for i in range(4):
    t.dot()
    t.forward(20)
    t.left(90)

t.forward(10)
t.left(90)
t.forward(10)
t.dot()

t.right(90)
t.forward(10)
t.pendown()
t.forward(width / 2)
t.stamp()

# Mass flux removed
t.penup()
t.seth(0)
t.goto(width / 4, height / 2)
for i in range(4):
    if i % 2 == 0:
        t.dot()
        t.forward(20)
        t.left(90)

t.forward(10)
t.left(90)
t.forward(10)
t.dot()

t.seth(270)
t.forward(10)
t.pendown()
t.forward(width / 2)
t.stamp()

# Mass flux produced
t.penup()
t.seth(0)
t.goto(3 * width / 4, height / 2)

t.forward(10)
t.left(90)
t.forward(10)
t.dot()

t.seth(90)
t.backward(10)
t.stamp()
t.pendown()
t.backward(width / 2)

# Mass flux out
t.penup()
t.goto(width + width / 2, height / 2)
t.seth(0)
for i in range(4):
    if i % 2 == 0:
        t.dot()
        t.forward(20)
        t.left(90)

t.forward(10)
t.left(90)
t.forward(10)
t.dot()

t.left(90)
t.backward(10)
t.stamp()
t.pendown()
t.backward(width / 2)
````

````python
showscreen()
clearscreen()

t = Turtle()

def draw_arithmetic_sequence(initial_length, diff):
    tracer(0)

    length = initial_length
    for i in range(10):
        for side in range(4):
            t.forward(length)
            t.left(90)
        length += diff

    tracer(1)

draw_arithmetic_sequence(1, 20)
````

````python
showscreen()
clearscreen()

t = Turtle()

def draw_geometric_sequence(initial_length, ratio):
    tracer(0)

    length = initial_length
    for i in range(10):
        for side in range(4):
            t.forward(length)
            t.left(90)
        length *= ratio

    tracer(1)

draw_geometric_sequence(2, 1.618)
````

````python
showscreen()
clearscreen()

t = Turtle()

# Set the grid dimensions
x_max = 200
x_min = -200
y_max = 200
y_min = -200

def draw_row(y, dot_size, spacing):
    # Iterate from left to right
    for x in range(x_min, x_max, spacing):
        # Update the turtle's position
        t.goto(x, y)
        # Draw a dot
        t.dot(dot_size)

def draw_grid(dot_size, spacing):
    tracer(0)

    t.penup()

    # Iterate from bottom to top
    for y in range(y_min, y_max, spacing):
        # Draw a row of dots
        draw_row(y, dot_size, spacing)
    
    tracer(1)

draw_grid(5, 10)
````

## Exploring Recursion

### 2-18

````python
showscreen()
clearscreen()

t = Turtle()

def draw_square_2(length):
    tracer(0)

    for side in range(4):
        t.forward(length)
        t.left(90)

    tracer(1)

def draw_arithmetic_sequence_r(length, diff):
    if length < 10:
        return
    draw_square_2(length)
    draw_arithmetic_sequence_r(length - diff, diff)

draw_arithmetic_sequence_r(175, 10)
````

````python
showscreen()

t.reset()

def draw_geometric_sequence_r(length, ratio):
    if length < 10:
        return
    draw_square_2(length)
    draw_geometric_sequence_r(length / ratio, ratio)

draw_geometric_sequence_r(175, 1.618)
````

### 2-19

````python
showscreen()
clearscreen()

t = Turtle()

def draw_stack_2(distance):
    if distance < 20:
        return
    else:
        draw_square_2(distance)
        draw_stack_2(distance + 20) # Oops!

draw_stack_2(175) # -> RecursionError: maximum recursion depth exceeded
````

````python
showscreen()
clearscreen()

t = Turtle()

def draw_tree(length, angle):
    if length < 10:
        t.dot()
        return
    else:
        # Draw the right branch
        t.forward(length)
        t.right(angle)
        draw_tree(length * 0.6, angle)
        t.left(angle)
        t.backward(length)
        # Draw the left branch
        t.forward(length)
        t.left(angle)
        draw_tree(length * 0.6, angle)
        t.right(angle)
        t.backward(length)

tracer(0)
t.left(90)
draw_tree(100, 60)
tracer(1)
````

````python
showscreen()
clearscreen()

t = Turtle()

def draw_sierpinski(length):
    if length < 5:
        return
    else:
        # Outer triangle
        t.forward(length)
        t.left(120)
        t.forward(length)
        t.left(120)
        t.forward(length)
        t.left(120)
        # Bottom left
        draw_sierpinski(length / 2)
        t.forward(length / 2)
        # Bottom right
        draw_sierpinski(length / 2)
        t.backward(length / 2)
        # Top
        t.left(60)
        t.forward(length / 2)
        t.right(60)
        draw_sierpinski(length / 2)
        t.left(60)
        t.backward(length / 2)
        t.seth(0)

draw_sierpinski(150)
````

````python
showscreen()
clearscreen()

t = Turtle()

# Set the grid dimensions
x_min = -200
x_max = 200
y_min = -200
y_max = 200

def draw_row_3(y, dot_size, spacing):
    # Iterate from left to right
    for x in range(x_min, x_max, spacing):
        # Set the turtle's drawing color
        if x ** 2 + y ** 2 < 160 ** 2:
            t.color("gray80")
        else:
            t.color("gray20")
        # Update the turtle's position
        t.goto(x, y)
        # Draw a dot
        t.dot(dot_size)

def draw_grid_3(dot_size, spacing):
    tracer(0)

    t.penup()

    # Iterate from bottom to top
    for y in range(y_min, y_max, spacing):
        # Draw a row of dots
        draw_row_3(y, dot_size, spacing)

    tracer(1)

draw_grid_3(5, 10)
````

### 2-24

````python
showscreen()
clearscreen()

t = Turtle()
t.hideturtle()

def follow_parabola(a, h, k, color):
    tracer(0)

    # Calculate the starting coordinates
    x_start = -200
    y_start = a * (x_start - h) ** 2 + k
    # Go to the start
    t.penup()
    t.goto(x_start, y_start)
    # Set the turtle's drawing color and pen size
    t.color(color)
    t.pensize(10)
    t.pendown()
    # Draw a parabola
    for i in range(400):
        x = x_start + i
        y = a * (x - h) ** 2 + k
        angle = t.towards(x, y)
        t.seth(angle)
        t.goto(x, y)
    
    tracer(1)

def follow_rainbow():
    for i in range(1, 10):
        # Compute the parabola's parameters
        a = -0.005
        h = 0
        k = i * 10
        # Style the parabola
        color = f"gray{i}0"
        # Draw the parabola
        follow_parabola(a, h, k, color)

follow_rainbow()
````

## Modeling Ecosystems

### 2-26

````python
showscreen()
clearscreen()

num_food = 10
num_animals = 20
step_size = 5

food = [Turtle() for i in range(num_food)]
animals = [Turtle() for i in range(num_animals)]
energy = [random.uniform(5, 10) for i in range(num_animals)]

for f in food:
    f.shape("circle")
    f.color(0.5, 0.5, 0.5)
    f.penup()
    x = random.uniform(-200, 200)
    y = random.uniform(-200, 200)
    f.teleport(x, y)

for a in animals:
    a.shape("turtle")
    a.showturtle()
    a.penup()
    x = random.uniform(-200, 200)
    y = random.uniform(-200, 200)
    a.teleport(x, y)

for i in range(100):
    for j in range(num_animals):
        if energy[j] >= 0:
            # Update position
            a = animals[j]
            x = a.xcor() + random.uniform(-step_size, step_size)
            y = a.ycor() + random.uniform(-step_size, step_size)
            angle = a.towards(x, y)
            a.seth(angle)
            a.goto(x, y)
            # Increase energy level
            for f in food:
                d = a.distance(f)
                if d < 20:
                    energy[j] += 1
                    food_x = random.uniform(-200, 200)
                    food_y = random.uniform(-200, 200)
                    f.teleport(food_x, food_y)
            # Decrease energy level
            energy[j] -= 0.1
            # Check energy level
            if energy[j] <= 0:
                energy[j] = 0
                a.hideturtle()
````

### 2-27

````python
tot_energy = 0
for e in energy:
    tot_energy += e
avg_energy = tot_energy / num_animals
print(f"The average in energy level in the population of {num_animals} animals is {avg_energy:.2}.")

dead_animals = 0
for i in range(num_animals):
    if energy[i] <= 0:
        dead_animals += 1

print(f"{dead_animals} animals died. So it goes.")
````
