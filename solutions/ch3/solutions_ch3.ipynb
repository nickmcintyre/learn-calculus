## Functions and Limits

### 3-1

We know that $ \frac{1}{3} = 0.333 \ldots $ so we can deduce the following:

$ \frac{1}{3} + \frac{1}{3} + \frac{1}{3} = 1 $

$ 0.333 \ldots + 0.333 \ldots + 0.333 \ldots = 1 $

$ 0.999 \ldots = 1 $

### 3-2

Even though $ f(5) $ doesn't exist, the image of nearby values such as $ f(4.999) $ and $ f(5.001) $ do exist. As $ x \to 5 $ from either direction, we can see that $ f(x) $ approaches $ 10 $, so $ \lim_{x \to 5} f(x) = 10 $.

### 3-3

Original functions:

$ f(x) = x + 5 $

$ g(x) = 2x $

Inverses:

$ f^{\, -1}(x) = x - 5 $

$ g^{-1}(x) = \frac{x}{2} $

Compositions:

$ f(f^{\, -1}(x)) = (x - 5) + 5 = x $

$ g(g^{-1}(x)) = 2(\frac{x}{2}) = x $

Composing a function with its inverse effectively "undoes" the original function, so the final image is the original argument $ x $.

### 3-4

```python
def f(x):
    return x ** 2 + 5

delta = 1
x_0 = 0
while abs(f(x_0 + delta) - f(x_0)) > 0.01:
    delta /= 2
print(f"δ = {delta}")
```

### 3-6

At the point of tangency, $ (1 ,\, 1) $, we know the slope is $ f^{\ -1} (1) = 2 \cdot 1 = 2 $. We can use this fact to derive the equation for the tangent line as follows:

$ m = \frac{\Delta y}{\Delta x} $

$ 2 = \frac{y - 1}{x - 1} $

$ 2 (x - 1) = \frac{y - 1}{x - 1} (x - 1) $

$ 2x - 2 = y - 1 $

$ 2x - 1 = y $

### 3-9

Let's start with the sum rule. First, we'll make a substitution for the sum of two functions and write its derivative using the limit definition.

$ h(x) = f(x) + g(x) $

$ \lim_{\delta \to 0} \frac{h(x \ + \ \delta) \ - \ h(x)}{\delta} $

Nothing new there. Now let's replace $ h(x) $ with $ f(x) + g(x) $ and see if we can find our two separate derivatives.

$ \lim_{\delta \to 0} \frac{[f(x \ + \ \delta) \ + \ g(x \ + \ \delta)] \ - \ [f(x) \ + \ g(x)]}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ + \ g(x \ + \ \delta) \ - \ f(x) \ - \ g(x)}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ - \ f(x) \ + \ g(x \ + \ \delta) \ - \ g(x)}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ - \ f(x)}{\delta} + \frac{g(x \ + \ \delta) \ - \ g(x)}{\delta} = f' + g' $

We can prove the difference rule using the same approach.

$ h(x) = f(x) - g(x) $

$ \lim_{\delta \to 0} \frac{h(x \ + \ \delta) \ - \ h(x)}{\delta} $

$ \lim_{\delta \to 0} \frac{[f(x \ + \ \delta) \ - \ g(x \ + \ \delta)] \ - \ [f(x) \ - \ g(x)]}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ - \ g(x \ + \ \delta) \ - \ f(x) \ + \ g(x)}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ - \ f(x) \ - \ g(x \ + \ \delta) \ + \ g(x)}{\delta} $

$ \lim_{\delta \to 0} \frac{f(x \ + \ \delta) \ - \ f(x)}{\delta} - \frac{g(x \ + \ \delta) \ - \ g(x)}{\delta} = f' - g' $

## Recursive Sequences

### 3-11

```python
def arithmetic_sequence(n, d):
    if n == 0:
        return 0
    else:
        # Return the next term: a(n) = a(n - 1) + d
        return arithmetic_sequence(n - 1, d) + d

arithmetic_sequence(3, 5)
```

### 3-12

```python
def factorial(n):
    """Computes n! = 1 * 2 * 3 * ... * n"""
    if n == 0:
        return 1
    else:
        # Return the next term: a(n) = a(n - 1) * n
        return factorial(n - 1) * n

factorial(5) # 5 * 4 * 3 * 2 * 1 = 120
```

```python
def bernoulli(n):
    """Computes a Bernoulli number"""
    if n == 0:
        return 1
    else:
        b = 0
        for k in range(n):
            num = factorial(n)
            den = factorial(n + 1 - k) * factorial(k)
            b += (num / den) * bernoulli(k)
        return -b

print(" n |     Bn")
print("-" * 16)
for i in range(21):
    print(f"{i:2} | {bernoulli(i):.8f}")
```

## Modeling Wind Power

### 3-24

$ v(z) = v_s \frac{\ln{\left(\frac{z}{z_0}\right)}}{\ln{\left(\frac{z_s}{z_0}\right)}} $

$ v(100) = 5 \frac{\ln{\left(\frac{100}{0.1}\right)}}{\ln{\left(\frac{10}{0.1}\right)}} $

```python
z = 100
z_0 = 0.1
z_s = 10
v_s = 5

v = v_s * math.log(z / z_0) / math.log(z_s / z_0)
print(f"v({z}) = {v:.2f} m/s")
```

### 3-25

$ P{(v)} = \begin{cases} 0 & v < 3.5 \\ 0.005832(v-3.5)^3 & 3.5 \leq v < 13 \\ 5 & 13 \leq v < 30 \\ 0 & 30 \leq v \end{cases} $

```python
def turbine_power(v):
    if v < 3.5:
        return 0
    if 3.5 <= v < 13:
        return 0.005832 * (v - 3.5) ** 3
    if 13 <= v < 30:
        return 5
    if v >= 30:
        return 0
```

```python
z = 100
z_0 = 0.1
z_s = 10
wind_speeds = [1, 5, 10]

print("   v   |   P(v)")
print("-" * 18)
for v_s in wind_speeds:
    v = v_s * math.log(z / z_0) / math.log(z_s / z_0)
    P = turbine_power(v)
    print(f"{v:6.2f} | {P:6.2f}")
```
