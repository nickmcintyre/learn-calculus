### 4-1

For the middle sum we have:

$ \Sigma_{i=0}^{2} f(x_i) = f(0.5) + f(1.5) + f(2.5) $

$ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \, = 0.5 + 1.5 + 2.5 $

$ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \, = 4.5 $

For the right sum we have:

$ \Sigma_{i=0}^{2} f(x_i) = f(1) + f(2) + f(3) $

$ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \,  = 1 + 2 + 3 $

$ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \,  = 6 $

### 4-2

Both the middle and right sums produce larger estimates than the left sum. The middle sum is the best estimate of the true integral; it happens to produce the exact value in this case.

$ \int dx \ 20x^2 = \frac{20x^3}{3} $

### 4-4

$ \int dx (a \pm b) = \int dx \, a \pm \int dx \, b $

Start from the sum and difference rules for derivatives:

$ A' = a $

$ B' = b $

$ (A \pm B)' = A' \pm B' $

Now integrate both sides:

$ \int dx \ (A \pm B)' = \int dx \  A' \pm \int dx \ B' $

$ a \pm b = a \pm b $ as required.

### 4-5

$ V = \int_{0}^{h} dy \ A(y)  $

$ \ \ \, = \frac{125}{3}  $

## Deriving Integration by Parts

### 4-10

$ \int dx \ f'(x) = \int dx \ g'(x) h(x) + \int dx \ g(x) h'(x) $

$ f(x) = \int dx \ g'(x) h(x) + \int dx \ g(x) h'(x) $

$ g(x) h(x) = \int dx \ g'(x) h(x) + \int dx \ g(x) h'(x) $

$ \int dx \ g(x) h'(x) = g(x) h(x) - \int dx \ g'(x) h(x)  $

### 4-11

$ g(x) = x \ \ \ \ \ h(x) = e^x $

$ g'(x) = 1 \ \ \ \ \ h'(x) = e^x $

$ \int dx \ x e^x = xe^x - \int dx \ 1 e^x $

$ \int dx \ x e^x = xe^x - e^x + C $

$ \frac{d}{dx} (xe^x - e^x + C) = xe^x + e^x - e^x $

$ \frac{d}{dx} (xe^x - e^x + C) = xe^x $

### 4-12

$ g(x) = x \ \ \ \ \ h(x) = -\cos{(x)} $

$ g'(x) = 1 \ \ \ \ \ h'(x) = \sin{(x)} $

$ \int dx \ x \sin{(x)} = x(-\cos{(x)}) - \int dx \ 1 (-\cos{(x)}) $

$ \int dx \ x \sin{(x)} = -x\cos{(x)} + \int dx \ \cos{(x)} $

$ \int dx \ x \sin{(x)} = -x\cos{(x)} + \sin{(x)} + C $

$ \frac{d}{dx} (-x\cos{(x)} + \sin{(x)} + C) = -\cos{(x)} + x\sin{(x)} + \cos{(x)} $

$ \frac{d}{dx} (-x\cos{(x)} + \sin{(x)} + C) = x\sin{(x)} $

### 4-13

$ g(x) = \sin{(x)} \ \ \ \ \ h(x) = e^x $

$ g'(x) = \cos{(x)} \ \ \ \ \ h'(x) = e^x $

$ \int dx \ \sin{(x)} e^x = \sin{(x)}e^x - \int dx \ (-\cos{(x)}) e^x $

$ \int dx \ \sin{(x)} e^x = \sin{(x)}e^x + \int dx \ \cos{(x)} e^x $

We could try to apply integration by parts again to $ \int dx \ \cos{(x)} e^x $, but we'd wind up in a similar position after doing so. It's an infinite loop!

$ S_c = \int_{0}^{Q_e} D(Q) - P_e $

```python
Qe = 90.91
Sp_e =  -0.003125 * Qe ** 2 + 0.57 * Qe
print(f"Sp ≈ {Sp_e:.2f} $/day")
```

Economic surplus:

## Modeling Wind Energy

### 4-21

Table 4-1 gives the antiderivative of an exponential function:

$ \int dv \ g'(v)e^{g(v)} = e^{g(v)} + C $

In the case of our exponential distribution we have $ g(v) = -v/7 $. Let's set up the definite integral for the pdf:

$ \int_{0}^{\infty} dv \frac{1}{7} e^{-v/7} $

Now, remember to be careful with that upper limit of integration:

$ -e^{-v/7}|_{0}^{\infty} $

$ [\lim_{v \to \infty} -e^{-v/7}] - (-e^{0}) $

$ 0 - (-1) $

$ 1 $

### 4-22

The expected value of a pdf is given by the following integral:

$ \bar{v} = \int_{0}^{\infty} dv \ v f(v) $

$ \bar{v} = \int_{0}^{\infty} dv \ v \frac{1}{7} e^{-v/7} $

We need to find the antiderivative to compute the definite integral. This is a good opportunity to apply integration by parts. Let's use the following setup:

$ \int dv \ g(v) h'(v) = g(v) h(v) - \int dv \ g'(v) h(v) $

$ g(v) = v \ \ \ \ \ \ \ h(v) = -e^{-v/7} $

$ g'(v) = 1 \ \ \ \ \ \ \ h'(v) = \frac{1}{7} e^{-v/7} $

$ \int dv \ v \frac{1}{7} e^{-v/7} = v (-e^{-v/7}) - \int dv \ 1 (-e^{-v/7}) $

$ \int dv \ v \frac{1}{7} e^{-v/7} = -ve^{-v/7} - (7e^{-v/7}) + C $

$ \int dv \ v \frac{1}{7} e^{-v/7} = e^{-v/7}(-v - 7) + C $

Now let's evaluate the definite integral, keeping an eye on that upper limit of integration:

$ e^{-v/7}(-v - 7) |_{0}^{\infty} $

$ [\lim_{v \to \infty} e^{-v/7}(-v - 7)] - e^{0}(-7) $

The product in the limit is a bit mysterious. One of the terms is $ e^{-v/7} $ which approaches $ 0 $ as $ v $ approaches $ \infty $. That means the limit is $ 0 $. Woohoo!

$ 0 - (-7) $

$ \bar{v} = 7 $

### 4-23

```python
def rayleigh(v):
    return (2 / 7.9) * (v / 7.9) * math.exp(-(v / 7.9) ** 2)

def v_x_rayleigh(v):
    return v * rayleigh(v)

v_low = 0
v_high = 150
v_bar = plt.left_sum(v_x_rayleigh, v_low, v_high)
print(f"v_bar = {v_bar:.3f} m/s")
```

### 4-24

The wind speed distributions have the same expected value and different shapes. The Rayleigh distribution has a clear peak and lower probabilities at wind speeds near $ 0 $.

### 4-25

```python
# Exponential pdf
def exponential(v):
    return (1 / 7.9) * math.exp(-v / 7.9)

# Rayleigh pdf
def rayleigh(v):
    return (2 / 7.9) * (v / 7.9) * math.exp(-(v / 7.9) ** 2)

# Probabilities of wind speeds in the interval [13, 30]
v_low = 13
v_high = 30
prob_exponential = plt.left_sum(exponential, v_low, v_high)
prob_rayleigh = plt.left_sum(rayleigh, v_low, v_high)

# Rated power of the wind turbine
MW = 5

# Electrical energy generated in MWh
hours = 8760
MWh_exponential = prob_exponential * hours * MW
MWh_rayleigh = prob_rayleigh * hours * MW

print("Distribution | MWh")
print("-" * 19)
print(f" Exponential | {int(MWh_exponential)}")
print(f" Rayleigh    | {int(MWh_rayleigh)}")
```

### 4-26

If we assume an exponential distribution of wind speeds, the turbine would generate about 2.5 times as much electricity as it would if we assumed a Rayleigh distribution. Even though these distributions have the same expected value of $ 7 \ \text{m/s} $, they lead to dramatically different electricity outputs over the course of a year.

### 4-28

```python
def line_x(t):
    return t - 200

def line_y(t):
    return 0

x_0 = line_x(0)
y_0 = line_y(0)
print(f"({x_0}, {y_0})") # (-200, 0)
```

```python
# Show the plot
plt.show()

# Prepare to draw
plt.clear()
path_turtle = Turtle()

def draw_path(xt, yt, num_steps, color=plt.GRAY60):
    """Draws a paramteric path"""
    tracer(10, 0)
    
    path_turtle.pencolor(color)
    path_turtle.pensize(1)
    
    x_0 = xt(0)
    y_0 = yt(0)
    path_turtle.teleport(x_0, y_0)
    for i in range(num_steps):
        x = xt(i)
        y = yt(i)
        path_turtle.goto(x, y)
    
    tracer(1, 10)

# Draw a parametric path
draw_path(line_x, line_y, 400)
```

### 4-30

Let's keep things simple and compute the error at each time step as the distance between the \verb|Turtle|'s $ y $-coordinate the $ y $-coordinate of the path:

$ e(t) = \verb|t.ycor()| - y(t) $

### 4-31

```python
# Show the plot
plt.show()

# Configure PID control
k_p = 0.03
k_i = 0.03
k_d = 0.002

err = 0
err_prev = 0
err_tot = 0

# Integrate
t_f = 0.4
dt = 0.001
N = int(t_f / dt)

for i in range(1, N):
    t_i = dt * i
    
    # Error
    err_prev = err
    err = path_turtle.ycor() - line_y(t_i)
    err_tot += dt * err
    
    # Proportional
    a_p = k_p * err

    # Integral
    a_i = k_i * err_tot

    # Derivative
    a_d = k_d * (err - err_prev) / dt

    # Send control signal
    a = a_p + a_i + a_d
    path_turtle.right(a)

    # Move forward
    path_turtle.forward(1)
```
