## Analyzing Vector Operations

### 5-5

Given two vectors of length $ N $, vector addition requires the following floating-point operations:

$ \mathbf{a} + \mathbf{b} = (a_0 + b_0, a_1 + b_1, \ldots, a_{N-1} + b_{N-1}) $

That's $ N $ additions, giving the operation the complexity $ O(N) $.

Given a vector of length $ N $, scalar multiplication requires the following floating-point operations:

$ s \mathbf{a} = (s a_0, s a_1, \ldots, s a_{N-1} ) $

That's $ N $ multiplications, giving the operation the complexity $ O(N) $.

### 5-6

```python
def list_add(a, b):
    """Add two vectors stored in lists"""
    N = len(a)
    c = []
    for i in range(N):
        c.append(a[i] + b[i])
    return c

def list_mult(s, a):
    """Scale a vector stored in a list"""
    N = len(a)
    b = []
    for i in range(N):
        b.append(s * a[i])
    return b

def arr_add(a, b):
    """Add two vectors stored in ndarrays"""
    c = a + b
    return c

def arr_mult(s, a):
    """Scale a vector stored in a ndarray"""
    b = s * a
    return b

s = 5
N = 1000
a_list = [i for i in range(N)]
b_list = [i for i in range(N)]
a_arr = np.arange(N)
b_arr = np.arange(N)

print("Vector addition")
%timeit list_add(a_list, b_list)
%timeit arr_add(a_arr, b_arr)

print("Vector scaling")
%timeit list_mult(s, a_list)
%timeit arr_mult(s, a_arr)
```

### 5-10

```python
def apply_gravity(r, mass_a, mass_b):
    """Computes the force of gravity between two bodies"""
    G = 6.67 * 10 ** (-11)
    r_mag = np.linalg.norm(r)
    strength = G * mass_a * mass_b / r_mag ** 2
    unit_vector = (1 / r_mag) * r
    return strength * unit_vector
```

```python
# Show the screen
showscreen()
clearscreen()

# Create the bodies
num_bodies = 3
body = [Turtle() for i in range(num_bodies)]

# Set the initial values
t_0 = 0
pos = []
pos.append(np.array([100.0, 0.0]))
pos.append(np.array([-100.0, 0.0]))
pos.append(np.array([1.0, 0.0]))
vel = []
vel.append(np.array([0.0, 500.0]))
vel.append(np.array([0.0, -500.0]))
vel.append(np.array([0.0, 0.0]))
acc = []
acc.append(np.array([0.0, 0.0]))
acc.append(np.array([0.0, 0.0]))
acc.append(np.array([0.0, 0.0]))

# Configure the Turtles
body[0].color(plt.GRAY60)
body[1].color(plt.GRAY20)
body[2].color(plt.WHITE)

# Set the parameters
mass = []
mass.append(10 ** 18)
mass.append(10 ** 18)
mass.append(10 ** 17)

# Configure the time step
t_f = 1
dt = 0.001
N = int(t_f / dt)

# Run Euler integration
for i in range(num_bodies):
    x, y = pos[i]
    body[i].teleport(x, y)

for i in range(1, N):
    # Compute forces and accelerations
    for j in range(num_bodies - 1):
        for k in range(j + 1, num_bodies):
            # Compute the force of gravity
            r_jk = pos[k] - pos[j]
            force_jk = apply_gravity(r_jk, mass[j], mass[k])
            force_kj = -force_jk
            # Compute the acceleration for bodies j and k
            acc_j = (1 / mass[j]) * force_jk
            acc_k = (1 / mass[k]) * force_kj
            # Update the accelerations for bodies j and k
            acc[j] += acc_j
            acc[k] += acc_k

    # Update velocity and position
    for j in range(num_bodies):
        vel[j] += dt * acc[j]
        pos[j] += dt * vel[j]
        x, y = pos[j]
        angle = body[j].towards(x, y)
        body[j].seth(angle)
        body[j].goto(x, y)

    # Reset acceleration
    for j in range(num_bodies):
        acc[j] *= 0
```

When a population's birth rate is greater than its death rate, the overall population grows. When the two rates are equal, the population size stays the same. When the death rate is larger than the birth rate, the population declines.

### 5-12

When a population's size greater than its carrying capacity, the term $ 1 - \frac{g}{N} $ will be negative and tend to decrease the overall population. When a population's size equals the carrying capacity, the population will remain at its current size. When the population size is less than its carrying capacity, the population will grow.

```python
# Set the initial values
g_0 = 5

# Set the parameters
kb = 1
kd = 0.5
N = 10

# Configure the time step
t = 2
dt = 0.01
num_steps = int(t / dt)

# Run Euler integration
g = g_0

for i in range(1, num_steps):
    g += dt * (kb - kd) * (1 - g / N) * g

print(f"g(0) = {g_0}")
print(f"g({t}) = {g}")
```

```python
# Set the initial values
S_0 = 997
I_0 = 3
R_0 = 0

# Set the parameters
b = 0.4
y = 0.04

# Configure the time step
t_f = 30
dt = 0.001
num_steps = int(t_f / dt)

# Run Euler integration
S = S_0
I = I_0
R = R_0

for i in range(1, num_steps):
    # Compute the derivatives
    dS_dt = -b * S * I
    dI_dt = b * S * I - y * I
    dR_dt = y * I
    # Update the population
    S += dt * dS_dt
    I += dt * dI_dt
    R += dt * dR_dt
print(" Population | Count")
print("-" * 20)
print(f"Susceptible | {int(S)}")
print(f"   Infected | {int(I)} ")
print(f"  Recovered | {int(R)}")
```

## Simulating Urban Air Pollution

### 5-14

Given:

$ k = 8.82 \times 10^{-3} \ \text{s}^{-1} $

$ [\text{NO}_2] = 10^{10} \ \text{molec/cm}^{3} $

Solution:

$ \frac{d[\text{NO}]}{dt} = k[\text{NO}_2] $

$ d[\text{NO}] = dt \ k[\text{NO}_2] $

$ \int d[\text{NO}] = \int dt \ k[\text{NO}_2] $

$ [\text{NO}] = k[\text{NO}_2]t + C $

$ [\text{NO}] = (8.82 \times 10^{-3})(10^{10})t + C $

$ [\text{NO}] = (8.82 \times 10^{7})t + C $

We can think of the constant of integration, $ C $, as the initial concentration of $ [\text{NO}] $.

$ [\text{NO}] = (8.82 \times 10^{7})t + [\text{NO}]_0 $

### 5-15

Given:

$ k = 6.59 \times 10^{-34} \ \text{cm}^3 / \text{s} $

$ [\text{O}] = 10 \ \text{molec/cm}^{3} $

$ [\text{O}_2] = 5.34 \times 10^{18} \ \text{molec/cm}^{3} $

Solution:

$ \frac{d[\text{O}_3]}{dt} = k[\text{O}][\text{O}_2] $

$ d[\text{O}_3] = dt \ k[\text{O}][\text{O}_2] $

$ \int d[\text{O}_3] = \int dt \ k[\text{O}][\text{O}_2] $

$ [\text{O}_3] = k[\text{O}][\text{O}_2]t + C $

$ [\text{O}_3] = (6.59 \times 10^{-34})(10)(5.34 \times 10^{18})t + C $

$ [\text{O}_3] = (3.52 \times 10^{-14})t + C $

We can think of the constant of integration, $ C $, as the initial concentration of $ [\text{O}_3] $.

$ [\text{O}_3] = (3.52 \times 10^{-14})t + [\text{O}_3]_0 $

### 5-16

Given:

$ k_a = 7.7 \times 10^{-12} \ \text{cm}^3 / \text{s} $

$ k_b = 8.82 \times 10^{-3} \ \text{s}^{-1} $

$ k_c = 6.59 \times 10^{-34} \ \text{cm}^3 / \text{s} $

Solution:

$ \frac{d[\text{NO}_2]}{dt} = k_a[\text{NO}][\text{RO}_2] $

$ \frac{d[\text{O}]}{dt} = k_b[\text{NO}_2] $

$ \frac{d[\text{O}_3]}{dt} = k_c[\text{O}][\text{O}_2] $

### 5-17

```python
# Set the initial values
O_0 = 0
O2_0 = 5.34 * 10 ** 18
O3_0 = 0
NO_0 = 10 ** 12
NO2_0 = 10 ** 10
RO_0 = 0
RO2_0 = 10 ** 11

# Set the parameters
ka = 7.7 * 10 ** 12
kb = 8.82 * 10 ** (-3)
kc = 6.59 * 10 ** (-34)

# Configure the time step
t_f = 1
dt = 0.0001
num_steps = int(t_f / dt)

# Run Euler integration
O = O_0
O2 = O2_0
O3 = O3_0
NO = NO_0
NO2 = NO2_0
RO = RO_0
RO2 = RO2_0

for i in range(1, num_steps):
    # Compute the derivatives
    dNO2_dt = ka * NO * RO2
    dO_dt = kb * NO2
    dO3_dt = kc * O * O2
    # Update the concentrations
    NO2 += dt * dNO2_dt
    O += dt * dO_dt
    O3 += dt * dO3_dt

print(" Chemical | molec/cm^3 ")
print("-" * 23)
print(" " * 6 + f"NO2 | {NO2:.2e}")
print(" " * 8 + f"O | {O:.2e} ")
print(" " * 7 + f"O3 | {O3:.2e}")
```

## Simulating Sync

### 5-18

```python
# Show the screen
showscreen()
clearscreen()

# Create the Turtles
osc_a = Turtle()
osc_b = Turtle()

# Configure the Turtles
osc_a.teleport(-50, 0)
osc_b.teleport(50, 0)

# Set the initial values
phase_a = 0
phase_b = 1.5

# Set the parameters
freq_a = 5
freq_b = 8
K = 4

# Configure the time step
t_f = 2
dt = 0.001
num_steps = int(t_f / dt)

# Run Euler integration
two_pi = 2 * math.pi

for i in range(1, num_steps):
    # Compute the phase change
    dphase_a = freq_a + (K / 2) * math.sin(phase_b - phase_a)
    dphase_b = freq_b + (K / 2) * math.sin(phase_a - phase_b)

    # Update the phases
    phase_a += dt * dphase_a
    phase_b += dt * dphase_b

    # Update the headings
    heading_a = phase_a * 180 / math.pi
    heading_b = phase_b * 180 / math.pi
    osc_a.seth(heading_a)
    osc_b.seth(heading_b)
```

### 5-19

$ K $ values greater than 3 or so cause the oscillators to begin synchronizing. 

### 5-20

$ \sin{(\theta_a - \theta_b)} $ represents the phase difference between two oscillators. When $ \theta_a \approx \theta_b $, the value of $ \sin{(\theta_a - \theta_b)} $ is nearly 0, so neither oscillator changes its phase by much. On the other hand, when $ \theta_b - \theta_a \approx 90^{\circ} $, the oscillators are completly out of phase. In this scenario, $ \sin{(\theta_a - \theta_b)} $ is nearly 1 and both oscillators change as much as possible.

### 5-21

When an oscillator's natural frequency is much higher than the coupling constant, its natural frequency dominates the rate at which it changes phase:

$ \dot{\theta}_a = \omega_a + \frac{K}{2} \sin{(\theta_b - \theta_a)} $

Since $ \sin{} $ oscillates between $ -1 $ and $ 1 $, the second term in the equation oscillates between $ -\frac{K}{2} $ and $ \frac{K}{2} $. A value of $ \omega_a $ that's $ 10 $ or $ 100 $ times greater than $ K $ will greatly diminish the effect of coupling with the other oscillator.

```python
# Show the screen
showscreen()
clearscreen()

# Create the Turtles
num_osc = 10
osc = [Turtle() for i in range(num_osc)]

# Set the initial values
freq = np.random.uniform(5, 10, num_osc)
phase = np.random.uniform(0, 2 * np.pi, num_osc)
d_phase = np.zeros(num_osc, dtype=np.float64)

# Configure the Turtles
x = np.linspace(-200, 200, num_osc)
for i in range(num_osc):
    osc[i].teleport(x[i], 0)

# Set the parameters
K = 4

# Configure the time step
t_f = 2
dt = 0.001
num_steps = int(t_f / dt)

# Run Euler integration
for i in range(1, num_steps):
    # Compute the phase changes
    for j in range(num_osc):
        d_phase[j] = freq[j] + (K / num_osc) * np.sum(np.sin(phase - phase[j]))
    
    # Update the phases
    phase += dt * d_phase

    # Update the headings
    heading = phase * 180 / np.pi
    for j in range(num_osc):
        osc[j].seth(heading[j])
```
