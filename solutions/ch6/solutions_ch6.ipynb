## Linear Algebra

### 6-1

The columns of matrix $ A $ are linearly independent, so they form a basis for $ R^3 $. Matrix $ B $, on the other hand, does not form a basis for $ R^3 $ because the third column is a linear combination of the first two:

$ \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix} + \begin{bmatrix} 4 \\ 5 \\ 6 \end{bmatrix} = \begin{bmatrix} 5 \\ 7 \\ 9 \end{bmatrix} $

### 6-2

The column space of a matrix is the set of all possible linear combinations of its column vectors. It is equivalent to codomain when we view matrix-vector multiplication as a function that maps a vector argument to a vector image.

### 6-3

Given $ \mathbf{a} = (3, \, 4) $, we have the following:

$ \mathbf{a} \cdot \hat{\mathbf{e}}_0 = 3 $

$ \mathbf{a} \cdot \hat{\mathbf{e}}_1 = 4 $

Computing the inner product of a vector $ \mathbf{a} $ and a standard basis vector "picks out" a component of $ \mathbf{a} $. 

### 6-4

We can show this directly by considering the column and row vectors of $ I $ when we multiply. Doing so means we're computing inner products using standard basis vectors, which has the effect of picking out elements of $ M $. Here's an example with $ 4 \times 4 $  square matrices:

$ M I = \begin{bmatrix} m_{0 , 0} & m_{0 , 1} \\ m_{1 , 0} & m_{1 , 1} \end{bmatrix} \begin{bmatrix} \hat{\mathbf{e}}_0 & \hat{\mathbf{e}}_1 \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} m_{0 , 0} \cdot 1 + m_{0 , 1} \cdot 0 & m_{0 , 0} \cdot 0 + m_{0 , 1} \cdot 1 \\ m_{1 , 0} \cdot 1 + m_{1 , 1} \cdot 0 & m_{1 , 0} \cdot 0 + m_{1 , 1} \cdot 1 \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} m_{0 , 0} & m_{0 , 1} \\ m_{1 , 0} & m_{1 , 1} \end{bmatrix} $

$ I M = \begin{bmatrix} \hat{\mathbf{e}}_0 \\ \hat{\mathbf{e}}_1 \end{bmatrix} \begin{bmatrix} m_{0 , 0} & m_{0 , 1} \\ m_{1 , 0} & m_{1 , 1} \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} 1 \cdot m_{0 , 0} + 0 \cdot m_{1 , 0} & 1 \cdot m_{0 , 1} + 0 \cdot m_{1 , 1} \\ 0 \cdot m_{0 , 0} + 1 \cdot m_{1 , 0} & 0 \cdot m_{0 , 1} + 1 \cdot m_{1 , 1} \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} m_{0 , 0} & m_{0 , 1} \\ m_{1 , 0} & m_{1 , 1} \end{bmatrix} $

The same pattern holds for multiplying $ N \times N $ matrices.

### 6-7

We compute the inner product of $ \mathbf{a} $ and $ \mathbf{b} $ as follows:

$ \mathbf{a} \cdot \mathbf{b} = (3, \, 4) \cdot (2, \, 0) $

$ \ \ \ \ \ \ \ \, = 3 \cdot 2 + 4 \cdot 0 $

$ \ \ \ \ \ \ \ \, = 6 $

Now let's compute the matrix-matrix product $ \mathbf{a}^T \mathbf{b} $:

$ \mathbf{a}^T \mathbf{b} = \begin{bmatrix} 3 & 4 \end{bmatrix} \begin{bmatrix} 2 \\ 0 \end{bmatrix} $

$ \ \ \ \ \ \ \, = 3 \cdot 2 + 4 \cdot 0 $

$ \ \ \ \ \ \ \, = 6 $

The operations are equivalent.

### 6-8

Let's compute the matrix-matrix product $ \mathbf{a} \mathbf{b}^t $:

$ \mathbf{a}^T \mathbf{b} = \begin{bmatrix} 3 \\ 4 \end{bmatrix} \begin{bmatrix} 2 & 0 \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} 3 \cdot 2 & 3 \cdot 0 \\ 4 \cdot 2 & 4 \cdot 0 \end{bmatrix} $

$ \ \ \ \ \ \ \, = \begin{bmatrix} 6 & 0 \\ 8 & 0 \end{bmatrix} $

This product, called the *outer product*, produces a matrix whose elements multiply every combination of the $ \mathbf{a} $ and $ \mathbf{b} $.

## Analyzing Vector and Matrix Operations

### 6-9

Let's analyze the Hadamard product of two $ N $-dimensional vectors, $ \mathbf{a} $ and $ \mathbf{b} $:

$ \mathbf{a} \odot \mathbf{b} = (a_0 \times b_0,\, a_1 \times b_1, \ldots, a_{N-1} \times b_{N-1})  $

The operation always requires $ N $ multiplications, so its complexity is $ O(N) $.

Next, let's analyze the inner product of two $ N $-dimensional vectors, $ \mathbf{a} $ and $ \mathbf{b} $:

$ \mathbf{a} \cdot \mathbf{b} = \sum_{i=0}^{N-1} = a_i \times b_i = a_0 \times b_0 +  a_1 \times b_1 + \ldots + a_{N-1} \times b_{N-1} $

The operation always requires $ N $ multiplications and $ N - 1 $ additions, so its complexity is $ O(N) $.

### 6-10

Let's analyze multiplication of a square matrix $ A $ by a scalar $ s $:

$ s A = s \begin{bmatrix} a_{0,0} & a_{0,1} & \ldots & a_{0,N-1} \\ a_{1,0} & a_{1,1} & \ldots & a_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ a_{N-1,0} & a_{N-1,1} & \ldots & a_{N-1,N-1} \end{bmatrix} $

$ s A = \begin{bmatrix} s \times a_{0,0} & s \times a_{0,1} & \ldots & s \times a_{0,N-1} \\ s \times a_{1,0} & s \times s \times a_{1,1} & \ldots & s \times a_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ s \times a_{N-1,0} & s \times a_{N-1,1} & \ldots & s \times a_{N-1,N-1} \end{bmatrix} $

The operation always requires $ N \times N = N^2 $ multiplications, so its complexity is $ O(N^2) $.

Next, let's analyze the addition of two square matrices $ A $ and $ B $:

$ A + B = \begin{bmatrix} a_{0,0} & a_{0,1} & \ldots & a_{0,N-1} \\ a_{1,0} & a_{1,1} & \ldots & a_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ a_{N-1,0} & a_{N-1,1} & \ldots & a_{N-1,N-1} \end{bmatrix} + \begin{bmatrix} b_{0,0} & b_{0,1} & \ldots & b_{0,N-1} \\ b_{1,0} & b_{1,1} & \ldots & b_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ b_{N-1,0} & b_{N-1,1} & \ldots & b_{N-1,N-1} \end{bmatrix} $

$ A + B = \begin{bmatrix} a_{0,0} + b_{0,0} & a_{0,1} + b_{0,1} & \ldots & a_{0,N-1} + b_{0,N-1} \\ a_{1,0} + b_{1,0} & a_{1,1} + b_{1,1} & \ldots & a_{1,N-1} + b_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ a_{N-1,0} + b_{N-1,0} & a_{N-1,1} + b_{N-1,1} & \ldots & a_{N-1,N-1} + b_{N-1,N-1} \end{bmatrix} $

The operation always requires $ N \times N = N^2 $ additions, so its complexity is $ O(N^2) $.

Next, let's analyze the Hadamard product of two square matrices $ A $ and $ B $:

$ A \odot B = \begin{bmatrix} a_{0,0} & a_{0,1} & \ldots & a_{0,N-1} \\ a_{1,0} & a_{1,1} & \ldots & a_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ a_{N-1,0} & a_{N-1,1} & \ldots & a_{N-1,N-1} \end{bmatrix} \odot \begin{bmatrix} b_{0,0} & b_{0,1} & \ldots & b_{0,N-1} \\ b_{1,0} & b_{1,1} & \ldots & b_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ b_{N-1,0} & b_{N-1,1} & \ldots & b_{N-1,N-1} \end{bmatrix} $

$ A \odot B = \begin{bmatrix} a_{0,0} \times b_{0,0} & a_{0,1} \times b_{0,1} & \ldots & a_{0,N-1} \times b_{0,N-1} \\ a_{1,0} \times b_{1,0} & a_{1,1} \times b_{1,1} & \ldots & a_{1,N-1} \times b_{1,N-1} \\ \vdots & \vdots & \ldots & \vdots \\ a_{N-1,0} \times b_{N-1,0} & a_{N-1,1} \times b_{N-1,1} & \ldots & a_{N-1,N-1} \times b_{N-1,N-1} \end{bmatrix} $

The operation always requires $ N \times N = N^2 $ multiplications, so its complexity is $ O(N^2) $.

### 6-11

Let's analyze multiplication of a square matrix $ A $ by a vector $ \mathbf{x} $:

$ A \mathbf{x} = \begin{bmatrix} a_{0,0} & a_{0,1} & \ldots & a_{0,N-1} \\ a_{1,0} & a_{1,1} & \ldots & a_{1,N-1} \\ \vdots & \vdots & \ddots & \vdots \\ a_{N-1,0} & a_{N-1,1} & \ldots & a_{N-1,N-1} \end{bmatrix} \begin{bmatrix} x_0 \\ x_1 \\ \vdots \\ x_{N-1} \end{bmatrix} $

$ A \mathbf{x} = \begin{bmatrix} a_{0,0} \times x_0 + a_{0,1} \times x_1 \ \ldots \ a_{0,N-1} \times x_{N-1} \\ a_{1,0} \times x_0 + a_{1,1} \times x_1 \ \ldots \ a_{1,N-1} \times x_{N-1} \\ \vdots \\ a_{N-1,0} \times x_0 + a_{N-1,1} \times x_1 + \ \ldots \ a_{N-1,N-1} \times x_{N-1} \end{bmatrix} $

The operation always requires $ N \times N = N^2 $ multiplications and $ N \times (N-1) = N^2 - N $ additions, so its complexity is $ O(N^2) $.

Next, let's analyze the multiplication of two square matrices $ A $ and $ B $:

$ A B = \begin{bmatrix} a_{0,0} & a_{0,1} & \ldots & a_{0,N-1} \\ a_{1,0} & a_{1,1} & \ldots & a_{1,N-1} \\ \vdots & \vdots & \ddots & \vdots \\ a_{N-1,0} & a_{N-1,1} & \ldots & a_{N-1,N-1} \end{bmatrix} \begin{bmatrix} b_{0,0} & b_{0,1} & \ldots & b_{0,N-1} \\ b_{1,0} & b_{1,1} & \ldots & b_{1,N-1} \\ \vdots & \vdots & \ddots & \vdots \\ b_{N-1,0} & b_{N-1,1} & \ldots & b_{N-1,N-1} \end{bmatrix} $

We can view this operation as several inner products. Each inner product multiplies a row vector of $ A $ are with a column vector of $ B $. I'll write the top row vector of $ A $ as $ \mathbf{a}_0 $ and the left column vector of $ B $ as $ \mathbf{b}_0 $. All together, we have:

$ A B = \begin{bmatrix} \mathbf{a}_0 \cdot \mathbf{b}_0 & \mathbf{a}_0 \cdot \mathbf{b}_1 & \ldots & \mathbf{a}_0 \cdot \mathbf{b}_{N-1} \\ \mathbf{a}_1 \cdot \mathbf{b}_0 & \mathbf{a}_1 \cdot \mathbf{b}_1 & \ldots & \mathbf{a}_1 \cdot \mathbf{b}_{N-1} \\ \vdots & \vdots & \ddots & \vdots \\ \mathbf{a}_{N-1} \cdot \mathbf{b}_0 & \mathbf{a}_{N-1} \cdot \mathbf{b}_1 & \ldots & \mathbf{a}_{N-1} \cdot \mathbf{b}_{N-1} \end{bmatrix} $


Each inner product requires $ N $ multiplications and $ N - 1 $ additions, giving us $ 2N - 1 $ flops. Since we compute $ N \times N = N^2 $ of these inner products, that means matrix-matrix multiplication requires $ N^2 \times (2N - 1) = 2N^3 - N^2 $ flops. So, its complexity is $ O(N^3) $.

## Simulating Swarms

### 6-17

```python
# Show the plot
plt.show()

# Prepare to draw
plt.clear()

# Configure the swarmalators
N = 200
swarmalators = []
freq = np.random.uniform(0, 2 * np.pi, N)
pos = np.random.uniform(-1, 1, [N, 2])
phase = np.random.uniform(0, 2 * np.pi, N)

# Create the swarmalators
tracer(10, 0)
for i in range(N):
    t = Turtle()
    t.shape("circle")
    t.speed(0)
    t.penup()
    c = phase[i] / (2 * np.pi)
    t.color((c, c, c))
    t.setpos(pos[i, :] * 100) # Scale coordinates for drawing
    swarmalators.append(t)
tracer(1, 10)
```

```python
# Show the plot
plt.show()

# Set the initial values
d_phase = np.zeros(N)
d_pos = np.zeros([N, 2])

# Set the parameters
S = 1
K = 0

# Configure the time step
t_f = 1
dt = 0.001
num_steps = int(t_f / dt)

# Run Euler integration
for t in range(1, num_steps):
    # Reset the derivatives
    d_phase[:] = 0
    d_pos[:, :] = 0
    # Synchronize each oscillator with the rest of the network
    for i in range(N):
        d_phase[i] = freq[i]
        for j in range(N):
            if i != j:
                # Calculate the vectors pointing between each pair of swarmalators
                pos_ij = pos[j, :] - pos[i, :]
        
                # Calculate the distance between each pair of swarmalators
                distance = np.linalg.norm(pos_ij)

                # Calculate the phase difference between each pair of swarmalators
                phase_ij = phase[j] - phase[i]

                # Sync phase
                d_phase[i] += (K / N) * np.sin(phase_ij) / distance
                
                # Sync position
                F_attract = (1 + S * np.cos(phase_ij)) * pos_ij / distance
                F_repulse = (1 / distance ** 2) * pos_ij
                d_pos[i, :] += F_attract - F_repulse

    # Update position and phase
    phase += dt * d_phase
    pos += dt * d_pos

    next_heading = np.degrees(phase)
    next_pos = 100 * pos # Scaled for drawing
    color = (next_heading % 360) / 360
    tracer(10, 0)
    for i in range(N):
        s = swarmalators[i]
        s.goto(next_pos[i])
        s.seth(next_heading[i])
        s.color(color[i], color[i], color[i])
    tracer(1, 10)
```
