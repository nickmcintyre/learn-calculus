### 8-2

$ \mathbf{v} \cdot \hat{\mathbf{e}}_0 = (3 ,\, 4) \cdot (1 ,\, 0) $

$ \ \ \ \ \ \ \ \ \ \, = 3 \cdot 1 + 4 \cdot 0 $

$ \ \ \ \ \ \ \ \ \ \, = 3 $

$ \mathbf{v} \cdot \hat{\mathbf{e}}_1 = (3 ,\, 4) \cdot (0 ,\, 1) $

$ \ \ \ \ \ \ \ \ \ \, = 3 \cdot 0 + 4 \cdot 1 $

$ \ \ \ \ \ \ \ \ \ \, = 4 $

$ \mathbf{v} \cdot \hat{\mathbf{v}} = (3 ,\, 4) \cdot (0.6 ,\, 0.8) $

$ \ \ \ \ \ \ \ \ \, = 3 \cdot 0.6 + 4 \cdot 0.8 $

$ \ \ \ \ \ \ \ \ \, = 5 $

$ \mathbf{v} \cdot \hat{\mathbf{v}} $ is the largest inner product.

### 8-4

The magnitudes $ | \mathbf{a} | $ and $ | \mathbf{b} | $ don't change, so the inner product is maximized when $ \cos{(\theta)} = 1 $. This occurs when $ \theta = 0 $, meaning the vectors point in the same direction.

### 8-5

Given $ f\,(x,\,y) = 3x + 4y $ and $ \nabla f = (3,\,4) $, we know that the unit vector $ \hat{\mathbf{u}} = (0.6, 0.8) $ gives the largest inner product $ \nabla f \cdot \hat{\mathbf{u}} $. We can see that $ \hat{\mathbf{u}} = \frac{\nabla f}{|\nabla f|} $.

### 8-6

#### Theorem

The gradient points in the direction of steepest ascent.

#### Proof

By definition, the inner product is maximized when vectors point in the same direction. Given two vectors $ \mathbf{a} $ and $ \mathbf{b} $:

$ \mathbf{a} \cdot \mathbf{b} = |\mathbf{a}||\mathbf{b}| \cos{(\theta)} $

Let $ \theta = 0 $:

$ \mathbf{a} \cdot \mathbf{b} = |\mathbf{a}||\mathbf{b}| \cos{(0)} $

$ \mathbf{a} \cdot \mathbf{b} = |\mathbf{a}||\mathbf{b}| $

Let $ \hat{\mathbf{u}} = \frac{\nabla f}{| \nabla f |} $ which is the unit vector in the direction of the gradient.

The projection of $ \nabla f $ in the direction of $ \hat{\mathbf{u}} $, $ \nabla f \cdot \hat{\mathbf{u}} $ is the projectiong with the largest magnitude. Put differently, the function $ f $ increases the most in the direction of its gradient, $ \hat{\mathbf{u}} $. The gradient points in the direction of steepest ascent. QED.

### 8-14

Consider the following error function and its partial derivatives:

$ E(w_0 ,\, w_1 ,\, w_2 ,\, w_3 ,\, b) =  \frac{1}{2} (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b)^2 $

$ E_{w_0} =  (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b) x_0 $

$ E_{w_1} =  (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b) x_1 $

$ E_{w_2} =  (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b) x_2 $

$ E_{w_3} =  (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b) x_3 $

$ E_{b} =  (w_0 x_0 + w_1 x_1 + w_2 x_2 + w_3 x_3 + b) 1 $

The derivatives are identical except for the term outside the parentheses. We can write the derivative conveniently in vector form using $ \mathbf{w} = (w_0 ,\, w_1 ,\, w_2 ,\, w_3 ) $ and $ \mathbf{x} = (x_0 ,\, x_1 ,\, x_2 ,\, x_3) $ like so:

$ E_{\mathbf{w}} =  (\mathbf{w} \cdot \mathbf{x} + b) \mathbf{x} $

$ E_{b} =  \mathbf{w} \cdot \mathbf{x} + b $
