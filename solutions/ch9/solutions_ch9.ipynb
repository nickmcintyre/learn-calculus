## Exploring the Discrete Fourier Transform

### 9-10

As $ T \to 0 $ and $ N \to \infty $, the DFT matrix becomes the continuous Fourier transform. The matrix has infinitely many rows and columns in the limit.

### 9-11

Given a device that samples signals with a given sampling rate $ T = 1 $,  the folding frequency is $ \xi_f = \frac{1}{2} $. Sampling once per second means that we can only detect signals that take $ 2 $ seconds to complete a cycle. If the signal cycles faster, then we won't be able to measure it.

### 9-12

The second equation gives us the following:

$ \hat{\mathbf{s}}_{N - k} = \Sigma_{n=0}^{N-1} \gamma_N^{(N-k)n} s_n $

$ \hat{\mathbf{s}}_{N - k} = \Sigma_{n=0}^{N-1} \gamma_N^{Nn}\gamma_N^{-kn} s_n $

$ \hat{\mathbf{s}}_{N - k} = \Sigma_{n=0}^{N-1} \gamma_N^{-kn} s_n $

Within the summation, the complex numbers $ \gamma_N^{kn} s_n $  and $ \gamma_N^{-kn} s_n $ have the same magnitude, so $ \hat{\mathbf{s}}_k $ and $ \hat{\mathbf{s}}_{N - k} $ appear mirrored when plotting the frequency spectrum of a real-valued signal such as audio.

## Analyzing the Fast Fourier Transform

```python
def fft(s):
    """Computes the discrete Fourier transform of a signal using a FFT"""
    N = len(s)
    if N == 1:
        return s
    else:
        # Recursively compute the DFT on each half of the signal
        a = fft(s[0::2])
        b = fft(s[1::2])
        # Compute the twiddle factors and twiddle
        y_N = np.exp(1j * 2 * np.pi / N)
        k = np.arange(N // 2)
        c = (y_N ** k) * b
        # Compute the DFT
        s_hat = np.zeros(N, dtype=np.complex128)
        s_hat[0:N // 2] = a + c
        s_hat[N // 2:N] = a - c
        return s_hat
```

### 9-13

The base case $ s_0 $ requires one comparison `N == 1` before returning. So, this level of recursion performs a total of $ N \times 1 $ flops and has $ O(N) $ complexity.

### 9-14

The subset of the signal $ (s_0,\,s_2) $ performs the following operations:

- `y_N = np.exp(1j * 2 * np.pi / N)` performs $ N \times 4 $ flops
- `c = (y_N ** k) * b` performs $ \frac{N}{2} \times 2 $ flops
- `s_hat[0:N // 2] = a + c` performs $ \frac{N}{2} \times 1 $ flops
- `s_hat[N // 2:N] = a - c` performs $ \frac{N}{2} \times 1 $ flops

That gives us $ 4N + (\frac{N}{2})2 + \frac{N}{2} + \frac{N}{2} = 6N $ flops at this level and has $ O(N) $ complexity.

### 9-15

The computation of the odd half of the signal is identical to the even half, so it also has $ O(N) $ complexity.

### 9-16

Both halves of the analysis have $ O(N) $ complexity, so the total complexity at the middle level of recursion is also $ O(N) $.

### 9-17

We're given a following signal with $ N = 4 $ samples:

$ \mathbf{s} = (s_0,\,s_1,\,s_2,\,s_3) $

Once the FFT reaches the final (top) level of recursion, we have the following values from previous calls to \verb|fft|:

$ \mathbf{a} = (a_0,\,a_1) = (s_0 + s_2,\, s_0 - s_2) $

$ \mathbf{b} = (b_0,\,b_1) = (s_1 + s_3,\, s_1 - s_3) $

We then use these to compute the DFT like so:

$ \gamma_4 = e^{\frac{j2\pi}{4}} = j $

$ \mathbf{c} = (\gamma_4^0,\,\gamma_4^1) \odot \mathbf{b} $

$ \ \ \ = (1,\,j) \odot (s_1 + s_3,\, s_1 - s_3) $

$ \ \ \ = (s_1 + s_3,\, js_1 - js_3) $

$ \hat{\mathbf{s}}_{[0,\,2]} = \mathbf{a} + \mathbf{c} $

$ \ \ \ \ \ \ \ \ \ = (s_0 + s_2 + s_1 + s_3,\, s_0 - s_2 + js_1 - js_3) $

$ \ \ \ \ \ \ \ \ \ = (s_0 + s_1 + s_2 + s_3,\, s_0 + js_1 - s_2 - js_3) $

$ \hat{\mathbf{s}}_{[2,\,4]} = \mathbf{a} - \mathbf{c} $

$ \ \ \ \ \ \ \ \ \ = (s_0 + s_2 - s_1 - s_3,\, s_0 - s_2 - js_1 + js_3) $

$ \ \ \ \ \ \ \ \ \ = (s_0 - s_1 + s_2 - s_3,\, s_0 - js_1 - s_2 + js_3) $

I'll write the full vector $ \hat{\mathbf{s}} $ as a column vector to help with readability:

$ \hat{\mathbf{s}} = \begin{bmatrix} s_0 + s_1 + s_2 + s_3 \\ s_0 + js_1 - s_2 - js_3 \\ s_0 - s_1 + s_2 - s_3 \\ s_0 - js_1 - s_2 + js_3 \end{bmatrix} $

This is the same result we get by applying the matrix-vector version of the DFT.

### 9-18

We can use our result from exercise 9-14 since the computation is the same; the top level of recursion has $ O(N) $ complexity.

### 9-19

Each level of recursion has $ O(N) $ complexity. If we perform $ N $ flops $ \log_2{N} $ times, that's a total of $ N \log_2{(N)} $ flops. So, the FFT has $ O(N \log_2{N}) $ complexity. You showed that matrix-vector multiplication has $ O(N^2) $ complexity in an exercise at the end of Chapter 6, so we can see why the FFT is a big performance win over the DFT.

## Analyzing Climate Data

### 9-23

````python
wind_speed = np.load("data/wrdb/wind_speed.npy")

num_hours = len(wind_speed)
min_speed = np.min(wind_speed)
max_speed = np.max(wind_speed)
print(f"{num_hours} hours of wind speed observations ranging from {min_speed:.2f} to {max_speed:.2f} m/s")
````

### 9-24

````python
# Show the plot
plt.show()

# Prepare to draw
plt.clear()
plt.draw_grid()
plt.draw_axes()
plt.draw_xlabel("t")
plt.draw_ylabel("s")

# Plot the signal
num_hours = 2 ** 16 # About 7.5 years
x = np.linspace(0, plt.X_MAX, num_hours)
y = 2 * wind_speed[:num_hours]
plt.scatter(x, y, color=plt.GRAY60, size=1)
````

````python
# Show the plot
plt.show()

# Compute the FFT
s_hat = sg.fft(wind_speed[:num_hours])

# Prepare to draw
plt.clear()
plt.draw_grid()
plt.draw_axes()
plt.draw_xlabel("f")
plt.draw_ylabel("s_hat")

# Plot the spectrum
plt.plot_spectrum(s_hat)
````

````python
# Compute the frequencies with the highest amplitudes
s_mag = np.abs(s_hat) # Magnitude of FFT
indices = np.argsort(s_mag, descending=True)
k = indices[0:10] # Top 10
T = 1 # Sampling period

print(" Frequency (1/hr) | Period (hr) |  s_hat")
print("-" * 42)
for i in range(len(k)):
    f = k[i] / (num_hours * T)
    per = 1 / f
    print(f"{f:17.4f} | {per:11.2f} | {int(s_mag[k[i]])}")
````

Wind speeds vary the most on time scales of about 1 hour, 1 day, and 1 year.
